Vector addition

```python
applicant_income = np.array(df['Applicant_Income'])
coapplicant_income = np.array(df['Coapplicant_Income'])
total_income = applicant_income+coapplicant_income
total_income
```

vector subtraction

```python
savings = total_income - np.array(df['Monthly_Expense'])
savings
```

Scalar Multiplication

```python
increased_savings = 1.1 * savings
increased_savings
```

Vector dot product

```python
cibil_score = np.array(df['CIBIL_Score'])
dot_product = np.dot(total_income,cibil_score)
dot_product
```

Matrix Creation

```python
financial_matrix = np.array(df[
    ['Applicant_Income','Coapplicant_Income','Loan_Amount','CIBIL_Score','Monthly_Expense']
]
)
financial_matrix
```

Matrix Addition

```python
A=financial_matrix[:2]
B=financial_matrix[2:4]
matrix_addition= A+B
matrix_addition
```

Matrix-Vector Multiplication

```python
A=np.array(df[['Applicant_Income','Coapplicant_Income','Monthly_Expense']].head(3))
vector=np.array(df['Debt_to_Income_Ratio'].head(3))
matrix_vector_multiplication=np.dot(A,vector)
matrix_vector_multiplication
```

Matrix-Matrix Multiplication

```python
A=np.array(df[["Applicant_Income","Coapplicant_Income"]])
B=np.array(df[["Loan_Amount","CIBIL_Score"]])
matrix_matrix_multiplication=np.dot(A,B.T)
matrix_matrix_multiplication
```

Find Rank in Numpy->

```python
A=np.array(df[["Applicant_Income","Coapplicant_Income"]])
rank = np.linalg.matrix_rank(A)
rank
```
